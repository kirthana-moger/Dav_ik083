```python
import pandas as pd
import numpy as np
```

```python
data={'Name':['Anu','Ravi','Priya','Kiran','Meena','Arjun'],
      'Age':[21,22,np.nan,23,21,np.nan],
      'Salary':[25000,np.nan,30000,280000,np.nan,32000],
      'Department':['CSE','ECE','CSE',np.nan,'ISE','ECE'],
      'City':['Mysore','Banglore','Mysore','Hubli','Banglore','Mysore']}
```

```python
df=pd.DataFrame(data)
print("Original data:")
print(df)
```

```python
print("\nMissing values:")
print(df.isnull())
```

```python
print("\n Number of missing values:")
print(df.isnull().sum())
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
print("\nAfter Mean Imputation:")
print(df)
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].median())
print("\nAfter Median Imputation:")
print(df)
```

```python
df['Department']=df['Department'].ffill()
print("\nAfter Forward Fill:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
dict2={
    "Name":["anu","reema","sanjay","ram"],
    "branch":["cse","eee","sce","ece"]
}
df1=pd.DataFrame(dict2)
print(df1)
encoder=LabelEncoder()
df1['Branch_Encoded'] = encoder.fit_transform(df1["branch"])
print(df1)
print(encoder.classes_)
```

```python
import pandas as pd
data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}
df = pd.DataFrame(data)
encoded_df = pd.get_dummies(df, columns=['Department'])
print(encoded_df)
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=['Department'],
    dtype=int
)
print(encoded_df)
```
